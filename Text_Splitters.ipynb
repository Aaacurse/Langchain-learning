```python
markdown_text='''# AI Documentation Guide

AI systems are used in many industries today. They help in automation, prediction, and decision-making.

## Introduction
Artificial Intelligence is a broad field that includes machine learning, deep learning, NLP, and robotics.

## Why AI is Important
AI improves efficiency and reduces manual effort in repetitive tasks.

### Benefits
- Faster processing
- Automation
- Better predictions
- Improved customer experience

### Drawbacks
- Bias in training data
- Privacy concerns
- Hallucinated answers

## Retrieval Augmented Generation (RAG)
RAG is a method where external documents are used to improve LLM responses.

### Steps in RAG
1. Load documents
2. Split into chunks
3. Convert chunks into embeddings
4. Store in vector database
5. Retrieve relevant chunks during query
6. Generate final answer using LLM

## Conclusion
Markdown documents are best split using MarkdownHeaderTextSplitter because it preserves section hierarchy.
'''
```

```python
python_code='''
import os
import json
from typing import List, Dict

class DocumentProcessor:
    def __init__(self, folder_path: str):
        self.folder_path = folder_path

    def list_files(self) -> List[str]:
        files = []
        for file in os.listdir(self.folder_path):
            if file.endswith(".txt"):
                files.append(file)
        return files

    def read_file(self, filename: str) -> str:
        full_path = os.path.join(self.folder_path, filename)
        with open(full_path, "r", encoding="utf-8") as f:
            return f.read()

    def clean_text(self, text: str) -> str:
        return text.replace("\n", " ").strip()

def generate_metadata(source: str, length: int) -> Dict:
    return {
        "source": source,
        "length": length
    }

def chunk_documents(texts: List[str], chunk_size: int = 300) -> List[str]:
    chunks = []
    for text in texts:
        for i in range(0, len(text), chunk_size):
            chunks.append(text[i:i+chunk_size])
    return chunks

def save_chunks(chunks: List[str], output_file: str):
    with open(output_file, "w", encoding="utf-8") as f:
        json.dump(chunks, f, indent=4)

if __name__ == "__main__":
    processor = DocumentProcessor("data")

    files = processor.list_files()
    print("Found files:", files)

    all_texts = []
    for file in files:
        content = processor.read_file(file)
        cleaned = processor.clean_text(content)
        all_texts.append(cleaned)

    chunks = chunk_documents(all_texts, chunk_size=200)
    save_chunks(chunks, "chunks.json")

    print("Total chunks saved:", len(chunks))
'''
```
