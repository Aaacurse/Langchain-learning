WebURL Loader

```python
from langchain_classic.document_loaders import WebBaseLoader
```

```python
url='https://www.amazon.in/Lapcare-WL-102-Wireless-Precision-Technology/dp/B0C81KYZ5J'
```

```python
web_loader = WebBaseLoader(
    web_paths=[url],
    header_template={
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36",
        "Accept-Language": "en-US,en;q=0.9",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8",
        "Connection": "keep-alive"
    }
)
```

```python
web_loaded=web_loader.load()
```

```python
web_loaded
```

Connect LLM

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_huggingface import HuggingFaceEndpoint,ChatHuggingFace
```

```python
llm=HuggingFaceEndpoint(
    repo_id="openai/gpt-oss-20b",
    task="conversational"
)
```

```python
model=ChatHuggingFace(llm=llm)
```

```python
response=model.invoke("what is the capital of france")
```

```python
response.content
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
from langchain_core.output_parsers import StrOutputParser
parser=StrOutputParser()
```

```python
prompt=PromptTemplate(
    template='''Based on the given information from the document or webpage text please answer the question.The document is given below:
    {information}.

    The question is {question}
    ''',
    input_variables=['information','question']
)
```

```python
chain=prompt|model|parser
```

```python
response=chain.invoke({'information':text_loaded,'question':'Tell me about this text'})
```

```python
print(response)
```
